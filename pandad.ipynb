```python
data = {"Name": ["Alice", "Bob", None, "Charlie", "David", "Alice"],
        "Age": [25, 32, 45, None, 28, 25],
        "Salary": [50000, 60000, 75000, 48000, None, 50000],
        "Department": ["HR", "Tech", "Tech", None, "Finance", "HR"]}
employees = pd.DataFrame(data)
print("Original data:")
display(employees) 
```

```python
clean_df = employees.copy()
clean_df["Age"] = clean_df["Age"].fillna(clean_df["Age"].mean())
```

```python
clean_df
```

```python
clean_df.dropna(subset=["Department"],inplace= True)
```

```python
clean_df
```

```python
clean_df.drop_duplicates(inplace=True)
print("\nCleaned data:")
display(clean_df)
```

```python
print("\nSalary stats:")
print(clean_df["Salary"].describe())
```

```python
Q1 = clean_df["Salary"].quantile(0.25)
Q3 = clean_df["Salary"].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
print("Lower: ",lower_bound)
print("Upper: ",upper_bound)
outliers = clean_df[(clean_df["Salary"] < lower_bound) | (clean_df["Salary"] > upper_bound)]
print("Outliers:\n", outliers["Salary"])
```

```python
import pandas as pd
import numpy as np

sales = pd.DataFrame({
"Date": pd.date_range("2023-01-01", periods=10),
"Product": np.random.choice(["Widget", "Gadget", "Tool"], 10),
"Units": np.random.randint(1, 20, 10),
"Unit_Price": np.random.uniform(10, 50, 10).round(2)
})
sales
```

```python
sales["Total"] = sales["Units"] * sales["Unit_Price"]
print("Sales data with total:")
display(sales)
```

```python
sales["Price_Category"] = sales["Unit_Price"].apply(
    lambda x: "High" if x > 30 else "Low")
print("\nWith price category:")
display(sales)
```

```python
sales["Units_Bin"] = pd.cut(sales["Units"], 
                           bins=[0, 5, 10, 15, 20],
                           labels=["low", "medium", "high", "very high"])
print("\nWith binned units:")
display(sales)
```

```python
pivot = sales.pivot_table(index="Product", 
                         columns="Price_Category", 
                         values="Total",  
                         aggfunc="sum"


)
print("\nPivot table:")
display(pivot)
```

```python
employees = pd.DataFrame({
"id": [1, 2, 3, 4],
"name": ["Alice", "Bob", "Charlie", "David"],
"dept_id": [101, 102, 101, 103]
})
departments = pd.DataFrame({
"dept_id": [101, 102, 104],
"dept_name": ["HR", "Engineering", "Marketing"]
})
salaries = pd.DataFrame({
"id": [1, 2, 3, 5],
"salary": [70000, 80000, 75000, 90000]
})
```

```python
emp_dept = pd.merge(employees, departments, on="dept_id")
print("Employees with departments (inner join):")
display(emp_dept)
```

```python
emp_dept_left = pd.merge(employees, departments, on="dept_id", how="left")
print("\nLeft join:")
display(emp_dept_left)
```

```python
emp_dept_right = pd.merge(employees, departments, on="dept_id", how="right")
print("\nRight join:")
display(emp_dept_right)
```

```python
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/titanic.csv"
titanic = pd.read_csv(url)

pivot1 = titanic.pivot_table(index="sex", 
columns="class", 
                            values="survived", 
                            aggfunc="mean")
print("Survival rate by sex and class:")
display(pivot1)
```

```python
pivot2 = titanic.pivot_table(index=["sex", "alone"],
                            columns="class",
                            values=["age", "fare"],
                            aggfunc={"age": "mean", "fare": ["mean", "count"]})
print("\nMulti-level pivot:")
display(pivot2)

pivot3 = titanic.pivot_table(index="sex",
                            columns="class",
                            values="survived",
                            aggfunc="count",
                            margins=True)
print("\nWith totals:")
display(pivot3)
```

```python
import pandas as pd
data = {"Name": ["alice smith", "bob jones", "carol wilson", "david brown"],
        "Email": ["alice@example.com", "bob@test.org", "carol@demo.net", "david@sample.com"],
        "Comments": ["Great service!", "Needs improvement.", "Awesome!!!", "Average."]}
df = pd.DataFrame(data)
print("Original text data:")
display(df)

df["Name"] = df["Name"].str.title()
df["Email_Domain"] = df["Email"].str.extract(r"@(.+)")
df["Comment_Length"] = df["Comments"].str.len()
df["Sentiment"] = df["Comments"].str.contains("Great|Awesome").map({True: "Positive", False : "Negative" })
print("\nAfter text processing:")
display(df)

from sklearn.feature_extraction.text import TfidfVectorizer
tfidf = TfidfVectorizer(max_features=3)
features = tfidf.fit_transform(df["Comments"])
tfidf_df = pd.DataFrame(features.toarray(), 
                      columns=tfidf.get_feature_names_out())
print("\nTF-IDF features:")
display(tfidf_df)
```
