```python
class Animal:
    def speak(self):
        return "Animal speaks"
    def speak1(self):
        return "Animal speaks\nWoof!"
animal = Animal()
print(animal.speak1())
```

```python
class Father:
    def skill(self):
        return "Gardening"
class Mother:
    def skill(self):
        return "Cooking"
class Child(Father, Mother):
    def skill(self):
        return Father.skill(self) + " and " + Mother.skill(self)
c = Child()
print(c.skill())
```

```python
class Grandparent:

    def house(self):
        return "Owns a big house"


class Parent(Grandparent):

    def car(self):
        return "Owns a car"


class Child(Parent):

    def bike(self):
        return "Owns a bike"


c = Child()
print(c.house())  # Inherited from Grandparent
print(c.car())  # Inherited from Parent
print(c.bike())  # Defined in Child
```

```python
class Parent:
    def property(self):
        return "Has land property"
class Son(Parent):
    pass
class Daughter(Parent):
    pass
s = Son()
d = Daughter()
print(s.property())
print(d.property())
```

```python
class A:

    def method_A(self):
        return "Method from A"
class B(A):
    def method_B(self):
        return "Method from B"
class C(A):
    def method_C(self):
        return "Method from C"
class D(B, C):
    def method_D(self):
        return "Method from D"
obj = D()
print(obj.method_A())
print(obj.method_B())
print(obj.method_C())
print(obj.method_D())
```

```python
class Animal:
    def __init__(self, name):
        self.name = name
        print("Animal created")
class Dog(Animal):
    def __init__(self, name, breed):
        super().__init__(name)
        self.breed = breed
        print("Dog created")
d = Dog("Buddy", "Labrador")
print(d.name)
print(d.breed)
```

```python
class Employee:
    company = "XYZ Ltd"
    def __init__(self, name):
        self.name = name
    @classmethod
    def get_company(cls):
        return cls.company
    @staticmethod
    def greet():
        return "Welcome to the company!"
print(Employee.get_company())
print(Employee.greet())
```

```python
class Book:
    def __init__(self, title, pages):
        self.title = title
        self.pages = pages
    def __str__(self):
        return f"Book: {self.title}, Pages: {self.pages}"
    def __len__(self):
        return self.pages
    def __mul__(self, other):
        return self.pages * other.pages
b1 = Book("Python", 300)
b2 = Book("Java", 400)
print(b1)
print(len(b1))
print("Total Pages:", b1 * b2)
```

```python
class Engine:
    def start(self):
        return "Engine started"
class Car:
    def __init__(self):
        self.engine = Engine()  # Composition

    def drive(self):
        return self.engine.start() + " → Car is moving"
c = Car()
print(c.drive())
```
