```python
import tkinter as tk
from tkinter import messagebox

def submit_feedback():
    name = name_entry.get().strip()
    feedback = feedback_entry.get().strip()
    rating = rating_var.get()

    if not name:
        messagebox.showwarning(
            "Missing Information",
            "Please enter your name."
        )
        name_entry.focus()
        return

    if not feedback:
        messagebox.showwarning(
            "Missing Information",
            "Please enter your feedback."
        )
        feedback_entry.focus()
        return

    if rating == "None":
        messagebox.showwarning(
            "Missing Information",
            "Please select a rating."
        )
        return

    result_label.config(
        text=(
            f"Student Name : {name}\n"
            f"Feedback     : {feedback}\n"
            f"Rating       : {rating}"
        )
    )

    messagebox.showinfo(
        "Submission Successful",
        f"Thank you, {name}!\n\n"
        "Your feedback has been submitted successfully."
    )

def clear_form():
    name_entry.delete(0, tk.END)
    feedback_entry.delete(0, tk.END)

    rating_var.set("None")

    result_label.config(
        text="Your submitted feedback will appear here."
    )

    name_entry.focus()

root = tk.Tk()

root.title("Student Feedback Management System")
root.geometry("650x650")
root.resizable(False, False)
root.configure(bg="#F4F6F7")


header_frame = tk.Frame(
    root,
    bg="#1F4E78",
    height=90
)
header_frame.pack(fill="x")


title_label = tk.Label(
    header_frame,
    text="STUDENT FEEDBACK SYSTEM",
    font=("Arial", 22, "bold"),
    fg="white",
    bg="#1F4E78"
)
title_label.pack(pady=(20, 5))


subtitle_label = tk.Label(
    header_frame,
    text="Please provide your valuable feedback",
    font=("Arial", 10),
    fg="#D6EAF8",
    bg="#1F4E78"
)
subtitle_label.pack()


form_frame = tk.Frame(
    root,
    bg="white",
    padx=30,
    pady=25
)
form_frame.pack(
    padx=40,
    pady=25,
    fill="both"
)

name_label = tk.Label(
    form_frame,
    text="Student Name",
    font=("Arial", 11, "bold"),
    bg="white",
    fg="#2C3E50"
)
name_label.pack(anchor="w")


name_entry = tk.Entry(
    form_frame,
    width=55,
    font=("Arial", 11),
    bg="#F8F9F9",
    fg="#17202A",
    relief="solid",
    bd=1
)
name_entry.pack(pady=(5, 20), ipady=6)


feedback_label = tk.Label(
    form_frame,
    text="Your Feedback",
    font=("Arial", 11, "bold"),
    bg="white",
    fg="#2C3E50"
)
feedback_label.pack(anchor="w")


feedback_entry = tk.Entry(
    form_frame,
    width=55,
    font=("Arial", 11),
    bg="#F8F9F9",
    fg="#17202A",
    relief="solid",
    bd=1
)
feedback_entry.pack(pady=(5, 20), ipady=6)

rating_label = tk.Label(
    form_frame,
    text="Overall Experience",
    font=("Arial", 11, "bold"),
    bg="white",
    fg="#2C3E50"
)
rating_label.pack(anchor="w", pady=(0, 8))


rating_var = tk.StringVar(value="None")


radio_frame = tk.Frame(
    form_frame,
    bg="white"
)
radio_frame.pack(anchor="w", pady=(0, 20))


radio_options = [
    ("Excellent", "Excellent"),
    ("Good", "Good"),
    ("Average", "Average")
]


for text, value in radio_options:

    radio_button = tk.Radiobutton(
        radio_frame,
        text=text,
        variable=rating_var,
        value=value,
        font=("Arial", 10),
        bg="white",
        fg="#2C3E50",
        activebackground="white",
        selectcolor="#D6EAF8"
    )

    radio_button.pack(
        side="left",
        padx=10
    )

button_frame = tk.Frame(
    form_frame,
    bg="white"
)
button_frame.pack(pady=5)


submit_button = tk.Button(
    button_frame,
    text="SUBMIT FEEDBACK",
    width=20,
    font=("Arial", 10, "bold"),
    bg="#27AE60",
    fg="white",
    activebackground="#229954",
    activeforeground="white",
    relief="flat",
    cursor="hand2",
    command=submit_feedback
)
submit_button.pack(
    side="left",
    padx=5,
    ipady=6
)


clear_button = tk.Button(
    button_frame,
    text="CLEAR",
    width=12,
    font=("Arial", 10, "bold"),
    bg="#E74C3C",
    fg="white",
    activebackground="#CB4335",
    activeforeground="white",
    relief="flat",
    cursor="hand2",
    command=clear_form
)
clear_button.pack(
    side="left",
    padx=5,
    ipady=6
)


# ------------------------------------------------------------
# Result Section
# ------------------------------------------------------------
result_title = tk.Label(
    root,
    text="Submission Details",
    font=("Arial", 12, "bold"),
    bg="#F4F6F7",
    fg="#1F4E78"
)
result_title.pack(pady=(5, 5))


result_label = tk.Label(
    root,
    text="Your submitted feedback will appear here.",
    width=65,
    height=5,
    font=("Arial", 10),
    bg="#D5F5E3",
    fg="#145A32",
    relief="solid",
    bd=1,
    justify="left",
    anchor="nw",
    padx=10,
    pady=10
)
result_label.pack(pady=(0, 15))


footer_label = tk.Label(
    root,
    text="© Student Feedback System | Python Tkinter",
    font=("Arial", 9),
    bg="#F4F6F7",
    fg="#7F8C8D"
)
footer_label.pack(
    side="bottom",
    pady=10
)


name_entry.focus()

root.mainloop()
```
