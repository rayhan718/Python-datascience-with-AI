```python
import numpy as np
import pandas as pd
np.random.seed(42)
num_students = 5
scores_array = np.random.randint(50, 101, size=(num_students, 3))
student_averages = np.mean(scores_array, axis=1)
student_names = ['Alice', 'Bob', 'Charlie', 'Diana', 'Ethan']
df = pd.DataFrame(scores_array, columns=['Math', 'Science', 'English'])
df['Student'] = student_names
df['Average'] = student_averages
df = df[['Student', 'Math', 'Science', 'English', 'Average']]

print("--- Combined NumPy & Pandas DataFrame ---")
print(df)
top_students = df[df['Average'] >= 75]
print("\n--- Students with Average Score >= 75 ---")
print(top_students)
print("\n--- Overall Subject Statistics ---")
summary_stats = pd.DataFrame({
    'Mean': df[['Math', 'Science', 'English']].mean(),
    'Max': df[['Math', 'Science', 'English']].max(),
    'Min': df[['Math', 'Science', 'English']].min()
})
print(summary_stats)
```
