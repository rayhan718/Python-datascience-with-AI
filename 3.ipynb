```python
def my_decorator(func):
    def wrapper():
        print("Before the function runs")
        func()
        print("After the function runs")
    return wrapper
@my_decorator
def say_hello():
    print("Hello")
    
say_hello()
        
```

```python
class Counter:
    def __init__(self,low,high):
        self.current=low
        self.high=high

    def __iter__(self):
        return self

    def __next__(self):
        if self.current >self.high:
            raise StopIteration
        else:
            self.current +=1
            return self.current -1
for num in Counter(1,5):
    print(num)
    
```

```python
def gen_numbers(n):
    for i in range(1,n+1):
        yield i
for num in gen_numbers(5):
    print(num)
```
