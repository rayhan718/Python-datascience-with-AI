```python
arr = np.array([[1,2,3],[4,5,6]])
print("Original arry(2x3:\n",arr)
print("\nFlattend (row-major):",arr.flatten())
print("\nFlattend(column-major):",arr.flatten('F'))
print("\nRaveld:",arr.ravel('F'))
```

```python
arr = np.array([[1, 2, 3], [4, 5, 6]])
print("Original array (2x3):\n", arr)
flat = arr.flatten()
print("\nFlattened (copy):", flat)
flat[0] = 100
print("Original after modifying flat:", arr[0,0])
raveled = arr.ravel()
print("\nRaveled (view):", raveled)
raveled[0] = 100
print("Original after modifying raveled:", arr[0,0])
```

```python
reshaped = arr.reshape(6)
print("\nReshaped to (6,):",reshaped)
reshaped[1]=200
print("Original after modifiying reshape:",arr[0,1])
```

```python
arr_1d = np.arange(12)
print("\n1d array:",arr_1d)
print("Reshaped to 3x4 :\n",arr_1d.reshape(3,4))
print("Reshaped to 2x3x2\n",arr_1d.reshape(3,4))
```
