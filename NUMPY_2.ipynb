```python
arr = np.array([[1, 2, 3], [4, 5, 6]])
print("Original array (2x3):\n", arr)
flat = arr.flatten()
print("\nFlattened (copy):", flat)
flat[0] = 100
print("Original after modifying flat:", arr[0,0])
raveled = arr.ravel()
print("\nRaveled (view):", raveled)
raveled[0] = 100
print("Original after modifying raveled:", arr[0,0])
```

```python
reshaped = arr.reshape(6)
print("\nReshaped to (6,):",reshaped)
reshaped[1]=200
print("Original after modifiying reshape:",arr[0,1])
```

```python
a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6], [7, 8]])
print(a)
print(b)
print("\nMatrix addition:\n", a + b)
print("\nMatrix multiplication:\n", np.matmul(a, b))
print("Dot product (same as matmul for 2D):\n", np.dot(a, b))
print("\nMatrix multiplication:\n", a @ b)  # Python 3.5+ operator    #prefered
print("\nElement-wise multiplication:\n", a * b)
print("\nMatrix transpose:\n", a.T)
```
