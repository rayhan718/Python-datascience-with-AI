```python
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author

    def display_book(self):
        print("Title:", self.title)
        print("Author:", self.author)

class Member:
    def __init__(self, name):
        self.name = name
        self.__borrowed_books = []

    def borrow_book(self, book):
        self.__borrowed_books.append(book)
        print(self.name, "borrowed", book.title)
    def display_books(self):
        print("Borrowed Books:")
        for book in self.__borrowed_books:
            print("-", book.title)

class StudentMember(Member):
    def __init__(self, name, roll_no):
        super().__init__(name)
        self.roll_no = roll_no

    def display_member(self):
        print("Student Name:", self.name)
        print("Roll No:", self.roll_no)


class FacultyMember(Member):
    def __init__(self, name, department):
        super().__init__(name)
        self.department = department

    def display_member(self):
        print("Faculty Name:", self.name)
        print("Department:", self.department)
book1 = Book("Python 101", "Guido")
book2 = Book("Effective Java", "Bloch")
student = StudentMember("Alice", 101)
faculty = FacultyMember("John", "Computer Science")
student.display_member()
student.borrow_book(book1)
student.display_books()

print()
faculty.display_member()
faculty.borrow_book(book2)
faculty.display_books()
```
